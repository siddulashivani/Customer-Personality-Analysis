```python
import pandas as pd
import numpy as np
```

```python
# Step 1 - Load and inspect the raw data

df = pd.read_csv('marketing_campaign.csv', sep='\t')
```

```python
# Step 2 - Capturing "before" stats for comparison later

before_rows = df.shape[0]
before_cols = df.shape[1]
before_nulls = df.isnull().sum().sum()
before_dupes = df.duplicated().sum()

print(f"Before Cleaning:")
print(f"Rows : {before_rows}")
print(f"Columns : {before_cols}")
print(f"Total missing values:{before_nulls}")
print(f"Duplicate rows: {before_dupes}")

print("\n Missing values by column:")
print(df.isnull().sum()[df.isnull().sum() > 0])
```

```python
# Step 3 - Handling missing values

median_income = df['Income'].median()
df['Income'] = df['Income']. fillna(median_income)

print("Missing Income values after fill:", df['Income'].isnull().sum())
```

```python
# Step 4 - Remove duplicates

duplicate_count = df.duplicated().sum()
print(f"Duplicate rows found: {duplicate_count}")

if duplicate_count > 0:
    df = df.drop_duplicates()
else:
    print("No duplicates to remove.")
```

```python
# Step 5 - Fixing data types

df['Dt_Customer'] = pd.to_datetime(df['Dt_Customer'], format='%d-%m-%Y')
print(df['Dt_Customer'].dtype)
```

```python
# Step 6 - Fixing invalid outlier values

reference_year = 2014 # latest enrollment year present in the dataset
implied_age = reference_year - df['Year_Birth']

outliers = df[implied_age > 100]
print(f"Rows with Implausible birth year (age > 100): {len(outliers)}")

df = df[implied_age <= 100].copy()
print(f"Rows remaining: {len(df)}" )
```

```python
# Step 7 - Cleaning inconsistent categories

print("Before cleanup:")
print(df['Marital_Status'].value_counts())

status_map = {'Alone': 'Single', 'Absurd': 'Single', 'YOLO': 'Single'}
df['Marital_Status'] = df['Marital_Status'].replace(status_map)

print("\nAfter cleanup:")
print(df['Marital_Status'].value_counts())
```

```python
# Step 8 - Droping columns with zero information

print("Z_CostContact Unique:", df['Z_CostContact'].unique())
print("Z_Revenue unique values:", df['Z_Revenue'].unique())

df = df.drop(columns = ['Z_CostContact', 'Z_Revenue'] )
```

```python
# Step 9 - Before/after summary

after_rows = df.shape[0]
after_nulls = df.isnull().sum().sum()
after_dupes = df.duplicated().sum()

print("Before and After Cleaning")
print(f"Rows: {before_rows} - {after_rows}")
print(f"Missing values: {before_nulls} - {after_nulls}")
print(f"Duplicate rows: {before_dupes} - {after_dupes}")
```

```python
# Step 10 - Saving the cleaned file

df.to_csv('marketing_campaign_cleaned.csv', index = False)
print("Saved File:", df.shape)
```
